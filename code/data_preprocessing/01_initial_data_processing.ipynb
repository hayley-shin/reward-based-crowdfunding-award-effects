## 1. 전체 프로젝트 찜 수 하나의 df로 합치기

```python
# 0-16600 index 프로젝트 찜 수 df 불러오기
likes_df1 = pd.read_csv('20221124_Wadiz_AllProjects_likes(0-16600)_df.csv')
# 16600-17950 index 프로젝트 찜 수 df 불러오기
likes_df2 = pd.read_csv('20221124_Wadiz_AllProjects_likes(16600-17950)_df.csv')

# likes_df1에서 16600 이후 행 제거
likes_df1 = likes_df1.drop(labels=range(16600, len(likes_df1)), axis=0)

# likes_df1랑 likes_df2 합치기
likes_df = pd.concat([likes_df1, likes_df2], axis = 0, ignore_index = True)
likes_df
```

```python
# csv로 저장
likes_df.to_csv('20221124_Wadiz_AllProjects_likes_df.csv', index=False, encoding='utf-8-sig')
```

## 2. 전체 프로젝트 디테일 하나의 df로 합치기

```python
# 0-16600 index 프로젝트 디테일 df 불러오기
detail_df1 = pd.read_csv('20221120_Wadiz_AllProjects_detail(0-16428)_df.csv')
# 16600-17950 index 프로젝트 디테일 df 불러오기
detail_df2 = pd.read_csv('20221120_Wadiz_AllProjects_detail(16429-17950)_df.csv')

# detail_df1에서 16428 이후 행 제거
detail_df1 = detail_df1.drop(labels=range(16429, len(detail_df1)), axis=0)

# detail_df1에서 detail_likes, campaignId, detailLink_x 칼럼 제거
detail_df1 = detail_df1.drop(['detail_likes', 'campaignId', 'detailLink_x'], axis=1)

# detail_df1랑 detail_df2 합치기
detail_df = pd.concat([detail_df1, detail_df2], axis = 0, ignore_index = True)
detail_df
```

```python
# csv로 저장
detail_df.to_csv('20221120_Wadiz_AllProjects_detail_df.csv', index=False, encoding='utf-8-sig')
```

## 3. likes_df랑 detail_df 합치기

```python
# 프로젝트 찜 수 df 불러오기
likes_df = pd.read_csv('20221124_Wadiz_AllProjects_likes_df.csv')
# 프로젝트 detail df 불러오기
detail_df = pd.read_csv('20221120_Wadiz_AllProjects_detail_df.csv')

# detail_df에 detail_likes 칼럼 추가하기
detail_df['detail_likes'] = likes_df['detail_likes']
# detail_df에 campaignId 칼럼을 첫 번째 칼럼으로 추가하기
detail_df.insert(0, 'campaignId', likes_df['campaignId'])

detail_df
```

```python
# csv로 저장
detail_df.to_csv('(20221122-20221124)Wadiz_AllProjects_detail_df_v1.csv', index=False, encoding='utf-8-sig')
```
